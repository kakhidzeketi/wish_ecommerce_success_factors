```python
wish_subset["has_urgency_banner"].fillna(0, inplace=True)
wish_subset["urgency_text"].fillna("none", inplace=True)
```

```python
wish_subset.isna().sum()
```

```python
wish_subset["origin_country"].value_counts()
```

```python
wish_subset["origin_country"].fillna("unknown", inplace = True)
wish_subset["product_color"].fillna("unknown", inplace = True)
wish_subset["product_variation_size_id"].fillna("unknown", inplace = True)
wish_subset["merchant_name"].fillna("unknown", inplace = True)
```
